加载 CFPS 成人数据，生成对数收入 log_income（加 1 避免收入为 0 时取对数出错），并删除教育程度 te4 为负值的缺失观测。

```python
%%stata
use ../datasets/cfps_adult, clear
gen log_income=log(p_income+1)
drop if te4<0
```

查看教育程度 te4 的分布，为按教育分组做方差分析做准备。

```python
%%stata
tab te4
```

方法一：用带组虚拟变量的回归实现单因素方差分析。i.te4 为每个教育水平生成虚拟变量，回归的整体 F 检验即检验各组均值是否全部相等；e(F) 与 e(rss) 分别保存 F 统计量与残差平方和。

```python
%%stata
// 用回归方法：
reg log_income i.te4
di "F=" e(F) " RSS=" e(rss)
```

方法二：直接用 anova 命令做单因素方差分析。可以看到两种方法给出的 F 统计量与 RSS 完全相同——单因素方差分析本质上就是对组虚拟变量回归的 F 检验，这是回归与方差分析统一性的一个具体体现。

```python
%%stata
// 用单因素方差分析
anova log_income te4
di "F=" e(F) " RSS=" e(rss)
```
