我们载入季度 GDP 数据并把 time 声明为时间变量。

```python
%%stata
use ../datasets/quarterlyGDP.dta, clear
tsset time
```

接着分别计算两种增长率并加上变量标签：同比增长率是与去年同季相比（滞后 4 期），可以消除季节性；环比增长率是与上一季度相比（滞后 1 期），更能反映短期动能。注意同比增速大致等于最近四个季度环比增速的累乘。

```python
%%stata
gen yoy_growth=(qgdp-L4.qgdp)/L4.qgdp
label variable yoy_growth "同比增长率"
gen pop_growth=(qgdp-L.qgdp)/L.qgdp
label variable pop_growth "环比增长率"
```

最后把两条增长率曲线画在同一张时间序列图上，便于比较两种口径下经济增长的波动。

```python
%%stata
tsline yoy_growth pop_growth
```
