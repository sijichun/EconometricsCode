```python
# 配置 PyStata：启动 Stata 并注册 %%stata 魔法命令
import os
os.chdir("../../code_in_notes/Chap.25")

import stata_setup
stata_setup.config("/opt/Stata", "mp")
```

我们载入季度 GDP 数据，并用 tsset 把 time 声明为时间变量，之后才能使用 L4. 这样的滞后算子。

```python
%%stata
use ../datasets/quarterlyGDP.dta, clear
tsset time
```

接着构造 GDP 的同比增长率：由于数据是季度频率，与一年前同一季度相比即滞后 4 期（L4.gdp），用（本期-去年同期）/去年同期得到同比增速，这样可以消除季节性。

```python
%%stata
gen gdp_growth=(gdp-L4.gdp)/l4.gdp
```

最后我们用 ac 命令考察增长率的自相关函数（autocorrelation function, ACF），lags(40) 表示计算并画出直到 40 阶的自相关系数，并给出 Bartlett 置信带，用以直观判断时间序列的记忆性与平稳性。

```python
%%stata
ac gdp_growth, lags(40)
```
