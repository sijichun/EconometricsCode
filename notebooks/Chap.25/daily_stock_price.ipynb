我们载入股票行情数据，并用 tsset 把数据声明为面板形式：stkcd 标识个体（股票），date 标识时间，这样 tsline 才能按时间正确连线。

```python
%%stata
use ../datasets/stock_price, clear
tsset stkcd date
```

我们用 tsline 画出代码为 600519 这只股票的收盘价日度时间序列图，直观观察价格水平的趋势与波动——价格序列通常呈现明显的非平稳特征。

```python
%%stata
tsline clsprc if stkcd == 600519
```
