我们生成 500 期的时间轴与标准正态白噪声 e，作为两个 AR(1) 过程共同的冲击序列。

```python
%%stata
clear
set obs 500
set seed 505
gen t=_n
gen e=rnormal()
tsset t
```

我们构造两个 AR(1) 过程：y1 的自回归系数为 0.9，y2 的自回归系数为 -0.9，两者使用相同的冲击 e。由于系数绝对值均小于 1，两个过程都平稳；但正系数使序列走势平滑、惯性强，负系数则使序列围绕均值上下交替震荡。

```python
%%stata
gen y1=0
replace y1=0.9*L.y1+e if t>1
gen y2=0
replace y2=-0.9*L.y2+e if t>1
```

画出 y1 的时间序列图：正的自回归系数使冲击的影响缓慢衰减，序列呈现持续的趋势性波动。

```python
%%stata
tsline y1
```

ac 命令计算并绘制 y1 的样本自相关函数（ACF）。平稳 AR(1) 且系数为正时，ACF 以 0.9^k 几何速度单调衰减，这是我们识别 AR 过程的重要特征。

```python
%%stata
ac y1
```

画出 y2 的时间序列图，可见负自相关带来的锯齿状波动。

```python
%%stata
tsline y2
```

y2 的 ACF 呈正负交替的几何衰减 ( (-0.9)^k )，这是负系数 AR(1) 的典型识别特征。

```python
%%stata
ac y2
```
