我们生成 500 期时间轴与标准正态白噪声 e，作为两个 MA 过程共同的冲击序列。

```python
%%stata
clear
set obs 500
set seed 505
gen t=_n
gen e=rnormal()
tsset t
```

构造两个移动平均过程：y1 为 MA(2)，y_t = e_t + 2 e_{t-1} + e_{t-2}；y2 为 MA(1)，y_t = e_t - 2 e_{t-1}。MA 过程是有限阶白噪声的加权平均，因此天然平稳。

```python
%%stata
gen y1=e+2*L.e+L2.e
gen y2=e-2*L.e
```

画出 y1 的时间序列图：MA 序列的波动比白噪声平滑，但没有 AR 那样的持续趋势。

```python
%%stata
tsline y1
```

绘制 y1 的 ACF。MA(q) 过程的自相关函数在 q 阶之后截尾（截断为 0）：y1 是 MA(2)，因此 ACF 在滞后 2 期之后落入置信区间内，这是识别 MA 阶数的关键特征。

```python
%%stata
ac y1
```

画出 y2 的时间序列图。

```python
%%stata
tsline y2
```

绘制 y2 的 ACF：作为 MA(1)，其一阶自相关系数显著为负（θ=-2 对应 ρ_1 = θ/(1+θ^2)），且滞后 2 期及以上截尾，再次印证 MA 的''q 阶截尾''性质。

```python
%%stata
ac y2
```
